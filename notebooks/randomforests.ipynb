```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from imblearn.ensemble import BalancedRandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv("../data/processed/dia_flights.csv")
print(df.info())
```

```python
#one classification model
y = df['15min_delay']
df.drop(columns=['15min_delay','Departure delay (Minutes)'],inplace=True)
x = df

x_train, x_test, y_train, y_test = train_test_split(
    x, y,
    test_size=0.2,
    random_state=43,
    stratify=y
)

clf = BalancedRandomForestClassifier(n_estimators=300,
                                     max_depth=None,
                                     min_samples_leaf=1,
                                     verbose=1,
                                     n_jobs=-1,
                                     random_state=43)
clf.fit(x_train, y_train)

y_pred = clf.predict(x_test)
```

```python
print(f"Accuracy score: \n {accuracy_score(y_test, y_pred)}")
print(f"Precision score: \n {precision_score(y_test, y_pred, average='weighted')}")
print(f"Recall score: \n {recall_score(y_test, y_pred, average='weighted')}")
print(f"f1 score: \n {f1_score(y_test,y_pred, average='weighted')}")
print(f"Confusion matrix: \n {confusion_matrix(y_test, y_pred)}")

#Get feature importances
importances = clf.feature_importances_
indices = np.argsort(importances)[::-1]
feature_names = df.columns[:-1]

for i in indices:
    print(f"{feature_names[i-1]}: {importances[i-1]:.4f}")
```
